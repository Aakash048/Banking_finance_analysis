```python
# Rerun the data-loading cell if df is undefined
df = pd.read_csv('/Users/aakasha/Documents/New project/Banking data analysis/Data/raw_data/bank_data.csv')

print(df.isnull().sum())
print(df.duplicated().sum())

df["bank_name"] = df["bank_name"].str.title()
df["city"] = df["city"].str.title()
df["transaction_type"] = df["transaction_type"].str.title()
df["channel"] = df["channel"].str.upper()
df["category"] = df["category"].str.title()
df["status"] = df["status"].str.upper()


df["amount_inr"] = pd.to_numeric(
    df["amount_inr"],
    errors="coerce"
)


# 7. Convert balance to numeric
df["balance_after_txn"] = pd.to_numeric(
    df["balance_after_txn"],
    errors="coerce"
)

df["transaction_id"] = df["transaction_id"].astype("string")
df["customer_id"] = df["customer_id"].astype("string")
df["account_id"] = df["account_id"].astype("string")
```

```python
df = df.fillna({ 'city': 'Unknown', 'channel': 'Unknown', 'category': 'Unknown', 'description': 'Not available' })
df['amount_inr'] = df['amount_inr'].fillna(df['amount_inr'].mean())
df['balance_after_txn'] = df['balance_after_txn'].fillna(df['balance_after_txn'].mean())
```
